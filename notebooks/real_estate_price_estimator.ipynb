# Real-Estate Property Price Estimator

## 1. Project Overview

### Objective
The objective of this project is to develop a machine learning system that estimates residential property prices using property characteristics and location-related features.

### Dataset
The project uses the provided training and test CSV files containing Indian residential property listings.

- **Training dataset:** 29,451 rows and 12 columns.
- **Test dataset:** 68,720 rows and 11 columns.
- **Target variable:** `TARGET(PRICE_IN_LACS)`.
- **Target unit:** Lakhs, as indicated by the dataset's target column and observed values.

### Machine Learning Models
Three regression algorithms will be evaluated:

1. Linear Regression — baseline model.
2. Random Forest Regressor — ensemble of decision trees.
3. XGBoost Regressor — gradient-boosted decision trees.

### Evaluation Metrics
Models will be evaluated using:

- **MAE:** Mean Absolute Error.
- **MSE:** Mean Squared Error.
- **RMSE:** Root Mean Squared Error.
- **R²:** Coefficient of determination.

### Project Workflow
1. Dataset auditing and exploratory data analysis (EDA).
2. Data cleaning and preprocessing.
3. Feature engineering and train-validation splitting.
4. Model training and comparison.
5. Final model selection and serialization.
6. Django backend integration.
7. React frontend integration and testing.

**Important:** The test dataset does not contain target prices. It will be reserved for generating predictions in the required submission format; model evaluation will use a validation set created from the training data.

## 6. Geographical Coordinate and Area Anomaly Analysis

### Objective
Verify the interpretation of the latitude and longitude columns and investigate unusually large reported property areas.

### Coordinate Validation
Geographical latitude ranges from −90° to +90°, while longitude ranges from −180° to +180°. A preliminary India bounding-box check uses latitude 6°–38° and longitude 68°–98°.

Because the initial results are inconsistent with many recognizable Indian locations, we will test whether the coordinate columns may have been reversed. We will not swap columns until the evidence supports that change.

### Property Area Validation
The reported area distribution contains extreme values that may affect regression performance. We will examine their associated prices, bedroom counts, and locations before deciding how to handle them.

**Important:** This step is diagnostic only. The original training data will remain unchanged.

```python
# 1. Test both possible coordinate interpretations.

def coordinate_summary(latitude_col, longitude_col, label):
    valid_india_box = (
        train[latitude_col].between(6, 38)
        & train[longitude_col].between(68, 98)
    )

    print(f"\n{label}")
    print("Latitude column:", latitude_col)
    print("Longitude column:", longitude_col)
    print("Rows inside broad India bounding box:", int(valid_india_box.sum()))
    print("Rows outside broad India bounding box:", int((~valid_india_box).sum()))


coordinate_summary("LATITUDE", "LONGITUDE", "Original coordinate interpretation")
coordinate_summary("LONGITUDE", "LATITUDE", "Swapped coordinate interpretation")

print("\nSample coordinate records:")
print(
    train[
        ["ADDRESS", "LONGITUDE", "LATITUDE"]
    ].sample(min(10, len(train)), random_state=42)
)

# 2. Investigate extremely large property areas.

area_anomalies = train.loc[
    train["SQUARE_FT"] > 100_000,
    [
        "POSTED_BY",
        "BHK_NO.",
        "BHK_OR_RK",
        "SQUARE_FT",
        "ADDRESS",
        "LONGITUDE",
        "LATITUDE",
        TARGET,
    ],
].sort_values("SQUARE_FT", ascending=False)

print("\nListings with reported area above 100,000 sq. ft.:")
print("Count:", len(area_anomalies))
print(area_anomalies)

# 3. Calculate price per square foot as a diagnostic.
# This is not a definitive validity test.

area_anomalies = area_anomalies.copy()
area_anomalies["PRICE_PER_SQ_FT_LAKHS"] = (
    area_anomalies[TARGET] / area_anomalies["SQUARE_FT"]
)

print("\nArea-anomaly price-per-square-foot diagnostics:")
print(
    area_anomalies[
        ["ADDRESS", "SQUARE_FT", TARGET, "PRICE_PER_SQ_FT_LAKHS"]
    ]
)
```

## 7. Data Cleaning Decisions

### 7.1 Coordinate Column Interpretation
The initial coordinate validation found zero records inside the preliminary India bounding box using the original column interpretation. Swapping the columns placed 29,193 records inside that box.

This strongly suggests that the coordinate column names are reversed. We will verify the pattern and, if confirmed, create correctly named latitude and longitude features during preprocessing rather than modifying the raw CSV files.

### 7.2 Property Area Anomalies
The dataset contains 72 listings with reported areas exceeding 100,000 sq. ft. Some have unusually low price-per-area ratios, suggesting possible data-quality issues.

However, area alone cannot establish whether a listing is invalid. We will use explicit, documented validation rules and evaluate their effect on model performance.

### 7.3 Reproducibility
The original CSV files will remain unchanged. All cleaning and feature engineering will be performed through a repeatable preprocessing pipeline, ensuring consistent transformations during model training and future predictions.

```python
# 1. Verify coordinate-column interpretation

swapped_latitude = train["LONGITUDE"]
swapped_longitude = train["LATITUDE"]

valid_swapped_coordinates = (
    swapped_latitude.between(6, 38)
    & swapped_longitude.between(68, 98)
)

print("Coordinate validation")
print("---------------------")
print("Valid latitude range after swapping:",
      int(swapped_latitude.between(-90, 90).sum()))
print("Valid longitude range after swapping:",
      int(swapped_longitude.between(-180, 180).sum()))
print("Inside preliminary India bounding box:",
      int(valid_swapped_coordinates.sum()))
print("Outside preliminary India bounding box:",
      int((~valid_swapped_coordinates).sum()))

print("\nRecords outside the preliminary India bounding box:")
print(
    train.loc[
        ~valid_swapped_coordinates,
        ["ADDRESS", "LONGITUDE", "LATITUDE", "SQUARE_FT", TARGET],
    ].head(20)
)

# 2. Quantify extreme-area records and their price ratios

price_per_sq_ft = train[TARGET] / train["SQUARE_FT"]

print("\nArea and price-ratio diagnostics")
print("--------------------------------")
print("Area > 10,000 sq. ft.:",
      int((train["SQUARE_FT"] > 10_000).sum()))
print("Area > 100,000 sq. ft.:",
      int((train["SQUARE_FT"] > 100_000).sum()))
print("Area > 1,000,000 sq. ft.:",
      int((train["SQUARE_FT"] > 1_000_000).sum()))

print("\nPrice per sq. ft. in lakhs — summary:")
print(price_per_sq_ft.describe(
    percentiles=[0.01, 0.05, 0.50, 0.95, 0.99]
).to_frame(name="PRICE_PER_SQ_FT_LAKHS"))

print("\nListings with the lowest price-per-area ratios:")
print(
    train.assign(PRICE_PER_SQ_FT_LAKHS=price_per_sq_ft)
    .nsmallest(15, "PRICE_PER_SQ_FT_LAKHS")[
        [
            "ADDRESS",
            "BHK_NO.",
            "SQUARE_FT",
            TARGET,
            "PRICE_PER_SQ_FT_LAKHS",
        ]
    ]
)
```

The original columns will remain unchanged in the raw dataset. The model pipeline will create corrected feature names only after validating the interpretation on both training and test data.

Address-derived locality and city candidates will also be retained as unverified categorical features because some suffixes appear inconsistent with the apparent location.

```python
# Validate the proposed coordinate swap in both datasets.
# This cell does not modify train or test.

INDIA_LAT_MIN, INDIA_LAT_MAX = 6.0, 38.0
INDIA_LON_MIN, INDIA_LON_MAX = 68.0, 98.0


def validate_coordinates(df, dataset_name):
    original_lat = df["LATITUDE"]
    original_lon = df["LONGITUDE"]

    swapped_lat = original_lon
    swapped_lon = original_lat

    valid_ranges = (
        swapped_lat.between(-90, 90)
        & swapped_lon.between(-180, 180)
    )

    inside_india_bounds = (
        swapped_lat.between(INDIA_LAT_MIN, INDIA_LAT_MAX)
        & swapped_lon.between(INDIA_LON_MIN, INDIA_LON_MAX)
    )

    print(f"\n{dataset_name}")
    print("-" * 45)
    print("Total rows:", len(df))
    print("Valid swapped latitude/longitude ranges:", int(valid_ranges.sum()))
    print("Inside preliminary India bounds:", int(inside_india_bounds.sum()))
    print("Outside preliminary India bounds:", int((~inside_india_bounds).sum()))
    print("Missing coordinate values:", int(df[["LONGITUDE", "LATITUDE"]].isna().any(axis=1).sum()))

    print("\nExamples outside preliminary India bounds:")
    print(
        df.loc[
            ~inside_india_bounds,
            ["ADDRESS", "LONGITUDE", "LATITUDE"]
        ].head(10)
    )


validate_coordinates(train, "TRAIN")
validate_coordinates(test, "TEST")

print("\nOriginal coordinate columns preserved:",
      {"LONGITUDE", "LATITUDE"}.issubset(train.columns)
      and {"LONGITUDE", "LATITUDE"}.issubset(test.columns))
```

```python
import numpy as np
import pandas as pd

train = pd.read_csv("../data/raw/train.csv")

train["SQUARE_FT"] = pd.to_numeric(
    train["SQUARE_FT"], errors="coerce"
)
train["TARGET(PRICE_IN_LACS)"] = pd.to_numeric(
    train["TARGET(PRICE_IN_LACS)"], errors="coerce"
)

train["AREA_RANGE"] = pd.cut(
    train["SQUARE_FT"],
    bins=[0, 2_000, 5_000, 10_000, 50_000, 100_000,
          1_000_000, np.inf],
    labels=[
        "0–2,000",
        "2,001–5,000",
        "5,001–10,000",
        "10,001–50,000",
        "50,001–100,000",
        "100,001–1,000,000",
        ">1,000,000",
    ],
    include_lowest=True,
)

area_price_summary = (
    train.groupby("AREA_RANGE", observed=False)
    .agg(
        Record_Count=("SQUARE_FT", "size"),
        Median_Area_SqFt=("SQUARE_FT", "median"),
        Median_Price_Lakhs=("TARGET(PRICE_IN_LACS)", "median"),
        Mean_Price_Lakhs=("TARGET(PRICE_IN_LACS)", "mean"),
        Max_Area_SqFt=("SQUARE_FT", "max"),
    )
    .reset_index()
)

print(area_price_summary)
```
